```python
import sqlite3
import pandas as pd

df = pd.read_csv('../data/cleaned_dataset.csv')
conn = sqlite3.connect('../data/analytics_db.db')
df.to_sql('sales_data', conn, if_exists='replace', index=False)
print("Data successfully imported into SQLite database!")
```

```python
query1 = "SELECT COUNT(*) AS total_rows FROM sales_data;"
pd.read_sql(query1, conn)
```

```python

query2 = """
SELECT Category, SUM(Sales) AS total_sales
FROM sales_data
GROUP BY Category
ORDER BY total_sales DESC
LIMIT 5;
"""
pd.read_sql(query2, conn)
```

```python
query3 = """
WITH AvgSales AS (
    SELECT AVG(Sales) AS avg_sale FROM sales_data
    )
    SELECT * FROM sales_data, AvgSales
    WHERE Sales > avg_sale
    LIMIT 10;
    """
pd.read_sql(query3, conn)
```
