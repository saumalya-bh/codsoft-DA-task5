```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv('../data/cleaned_dataset.csv')

# 1. Seaborn Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x='Category', y='Sales', data=df, hue='Category', palette='Set2', legend=False)
plt.title('Sales Distribution by Category')
plt.savefig('../reports/category_boxplot.png', bbox_inches='tight')
plt.show()

# 2. Plotly Interactive Donut Chart.
fig = px.pie(df, names='Category', values='Sales', hole=0.4, title='Sales Split by Category')
fig.write_html('../reports/interactive_sales_pie.html')
fig.show()

# 3. Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title('Correlation Matrix')
plt.tight_layout()
plt.savefig('../dashboards/seaborn_heatmap.png')
plt.show()

# 4. Boxen Plot (Distribution of Sales by Category)
plt.figure(figsize=(8, 5))
sns.boxenplot(data=df, x='Category', y='Sales')
plt.title('Sales Distribution by Category')
plt.tight_layout()
plt.savefig('../dashboards/seaborn_boxenplot.png')
plt.show()
```
